```python


products = {
    "Milk": 50,
    "Bread": 30,
    "Eggs": 60,
    "Rice": 80,
    "Oil": 120
}

cart = {}
total = 0

print("Welcome to Instamart 🛒")
print("Available Products:")
for item, price in products.items():
    print(f"{item} - ₹{price}")

while True:
    choice = input("\nEnter product name (or 'done' to finish): ").capitalize()
    if choice == "Done":
        break
    if choice in products:
        qty = int(input(f"Enter quantity of {choice}: "))
        cart[choice] = cart.get(choice, 0) + qty
        total += products[choice] * qty
    else:
        print("Product not available!")

print("\n🧾 Final Bill")
for item, qty in cart.items():
    print(f"{item} x {qty} = ₹{products[item] * qty}")
print(f"Total Amount: ₹{total}")
```
